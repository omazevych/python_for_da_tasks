## Завдання 3: Box Plot порівняння - Pandas vs Seaborn

**Завдання:**
Побудуйте box plot для кількості погодинних оренд велосипедів за погодними умовами з Pandas та Seaborn.

**Функція Seaborn: `sns.boxplot()`**

Можна побудувати окремо два графіки. Але для тих, хто хоче складніше - побудуйте ці 2 графіки на 1 фігурі.

Просунуте доповнення:
- підпишіть погодні умови їх інтерпретацією з опису даних в обох графіках

**Дайте відповідь на питання:**
- Яка візуальна різниця між Pandas та Seaborn бокс-плотами?

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
weather_names = { 1: 'Ясно', 2: 'Туман', 3: 'Легкий дощ', 4: 'Сильний дощ'}
df['weather_n'] = df['weather'].map(weather_names)

# Pandas
df.boxplot(column='count', by='weather_n', ax=ax[0], grid=False)
ax[0].set_title('Кількість погодинних оренд за погодою: Pandas', fontweight='bold')
ax[0].set_xlabel('Погодні умови', fontweight='bold')
ax[0].set_ylabel('Кількість оренд', fontweight='bold')

# Seaborn
sns.boxplot(data=df, x='weather_n', y='count', ax=ax[1])
ax[1].set_title('Кількість погодинних оренд за погодою: Seaborn', fontweight='bold')
ax[1].set_xlabel('Погодні умови', fontweight='bold')
ax[1].set_ylabel('Кількість оренд', fontweight='bold')

plt.suptitle('')
plt.tight_layout()
plt.show()
```

- Яка візуальна різниця між Pandas та Seaborn бокс-плотами?
*Seaborn більш естетична версія графіка, до того ж у цій версії дані розташовані за розподілом від найбільшого до найменшого, у Pandas дані розташовані в алфавітному порядку за осею х.*

## Завдання 4: Heatmap кореляційної матриці

**Завдання:**
Створіть із Seaborn кореляційну матрицю з забарвленням heatmap (колір відповідає величині значення в клітинці) числових змінних в наших даних з анотаціями значень.

**Дайте відповіді на питання по графіку:**
1. Які змінні найсильніше корелюють з загальною кількістю оренди (count)?
2. Яка кореляція між temp та atemp? Чому?

```python
correlation = df[['season', 'holiday', 'workingday', 'weather', 'temp', 'atemp', 'humidity', 'windspeed', 'casual', 'registered', 'count']].corr()

plt.figure(figsize=(12, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, linewidths=0.7, fmt="0.3f")
plt.title('Кореляційна матриця оренди велосипедів')
plt.tight_layout()
plt.show()
```
